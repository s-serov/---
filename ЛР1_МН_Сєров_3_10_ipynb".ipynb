### Завдання 27
Дано матриці:

```text
A = [[1, 2],
     [3, 4]]

B = [[2, 0],
     [1, 2]]
```

Обчисліть:
1. поелементний добуток `A * B`;
2. матричний добуток `A @ B`.

Поясніть у коментарі до коду, чим відрізняються ці операції.

````python
A = np.array([[1, 2], [3, 4]])
B = np.array([[2, 0], [1, 2]])

# Поелементний добуток: множаться елементи на однакових позиціях (A[i,j] * B[i,j])
elem_mult = A * B

# Матричний добуток: множення рядків першої матриці на стовпці другої за правилами лінійної алгебри
mat_mult = A @ B

print(elem_mult)
print(mat_mult)
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
A = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
], dtype=float)

mean = np.mean(A, axis=0)
std = np.std(A, axis=0)

z = (A - mean) / std
print(z)
````

### Завдання 37
Створіть квадратну матрицю `5 × 5` із випадкових цілих чисел від `1` до `20`.

Виконайте:
1. знайдіть суму головної діагоналі;
2. знайдіть суму побічної діагоналі;
3. знайдіть максимальний елемент кожного рядка;
4. знайдіть мінімальний елемент кожного стовпця;
5. замініть усі елементи, менші за середнє значення всієї матриці, на `0`.

````python
mat = np.random.randint(1, 21, size=(5, 5))
print(mat)

# Суми діагоналей
main_diag_sum = np.trace(mat)
anti_diag_sum = np.trace(np.fliplr(mat))

# Екстремуми по осях
row_max = mat.max(axis=1)
col_min = mat.min(axis=0)

# Заміна елементів, менших за середнє
mean_val = mat.mean()
mat[mat < mean_val] = 0

print(main_diag_sum)
print(anti_diag_sum)
print(row_max)
print(col_min)
print(mat)
````

Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
